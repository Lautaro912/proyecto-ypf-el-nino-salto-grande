Seguimos...

Durante El Niño, los vientos alisios se debilitan. Al no haber viento que empuje el agua hacia el oeste, la masa de agua cálida del Pacífico central y occidental fluye de regreso hacia las costas de Sudamérica (Perú, Colombia, Ecuador)

# Introducción
¿Qué tiene que ver el océano con la energía que usamos en casa? A primera vista, nada. Pero si se sigue el recorrido del agua, la conexión aparece.

El Pacífico funciona como una enorme reserva de calor. El agua tiene una capacidad térmica muy alta, así que el océano se calienta y se enfría mucho más lento que la atmósfera. Esa lentitud le permite guardar señales térmicas durante meses. Es, en cierto modo, una memoria del clima.

De esa interacción entre océano y atmósfera surge ENSO —El Niño–Oscilación del Sur—, con sus fases El Niño, La Niña y neutral. En condiciones normales, los vientos alisios empujan las aguas cálidas hacia el Pacífico occidental. Durante El Niño esos vientos se debilitan y el calor se redistribuye: la superficie del Pacífico ecuatorial central y oriental queda más cálida de lo habitual.

Eso mueve la circulación atmosférica y las lluvias en Sudamérica. Pero no de una manera uniforme. Depende de la región y de la época del año. En algunos lugares llueve más; en otros, menos. Por eso no tiene sentido decir que El Niño siempre reduce el agua disponible para generar electricidad. La relación es más complicada.

La conexión con la energía está en la cuenca. Lo que llueve, lo que queda en el suelo y, en algunas regiones, lo que se derrite de la nieve terminan definiendo cuánta agua llega al embalse. Después entran en juego las decisiones de operación. Pero el punto de partida es el agua disponible, y por eso nos interesa entender cómo varía.

En este proyecto vamos a trabajar con el caudal afluente, es decir, el agua que entra al embalse. Es una variable más directa que el nivel del embalse, que también depende de cómo se opere. Implicación para el proyecto.
La estacionalidad deberá considerarse al explorar la relación entre caudal y los índices ENSO. Estos gráficos describen el comportamiento del caudal, pero todavía no demuestran una asociación con ONI, RONI o SOI.

```python
# Valores atípicos según el criterio de 1,5 × IQR
def contar_atipicos(serie):
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    return (
        (serie < limite_inferior) |
        (serie > limite_superior)
    ).sum()


# Resumen por mes calendario, sin modificar df_salto_grande
display(
    df_salto_grande["valor"]
    .groupby(
        pd.to_datetime(
            df_salto_grande["timestart"], errors="raise"
        ).dt.month
    )
    .agg(
        cantidad="count",
        media="mean",
        mediana="median",
        minimo="min",
        maximo="max",
        atipicos=contar_atipicos
    )
    .rename_axis("mes")
    .round(2)
)
```

Se agruparon los caudales medios mensuales por mes calendario: todos los eneros, todos los febreros y así sucesivamente. Para cada grupo se calculó la cantidad de observaciones, la media, la mediana, el mínimo, el máximo y el número de valores atípicos. Los caudales se expresan en m³/s.

La comparación entre media y mediana permite evaluar la influencia de los caudales extremos. La media es más sensible a valores muy elevados, mientras que la mediana representa el centro de la distribución con menor influencia de estos extremos.

Se consideraron atípicas las observaciones inferiores a Q1 - 1,5 * IQR o superiores a Q3 + 1,5 * IQR, donde IQR = Q3 - Q1. Los límites se calcularon por separado para cada mes, para considerar las diferencias estacionales.

La columna `atipicos` indica cuántos promedios mensuales quedaron fuera de esos límites. Por ejemplo, un valor de 2 en enero representa dos eneros de distintos años, no dos días de crecida.

Los valores señalados se conservan: pueden representar situaciones hidrológicas reales y no necesariamente errores. Su identificación orienta la revisión de eventos excepcionales, pero no demuestra una relación con ENSO ni constituye, por sí sola, una clasificación de sequías o crecidas.

# Conceptos importantes:    


**1 - El Océano como Sumidero Térmico y Memoria del Clima**, explicamos la inercia térmica del océano.


> **La física del sumidero hídrico**: Por cada 100 Joules de exceso de energía acumulados en el sistema Tierra por el calentamiento global, más de 90 Joules son absorbidos por el océano. El océano es el gran amortiguador del clima debido a que el calor específico del agua es extremadamente alto (~4.000 J/(kg·K)).

> **Océano vs. Atmósfera**: En contraste con la atmósfera, que tiene muy poca masa y responde casi de inmediato a las perturbaciones energéticas sin capacidad de retener calor a largo plazo, el océano posee una masa colosal en contacto directo con el aire. Los procesos de mezcla y convección permiten que el calor superficial penetre lentamente a profundidad, dándole al océano una inercia térmica gigante.

"Lo importante" > Mientras que la atmósfera olvida rápidamente un forzante térmico, el océano aún lo recuerda. Esa "memoria oceanográfica" almacena la señal térmica de anomalías de temperatura durante meses, lo que nos da la ventana temporal perfecta para predecir variables terrestres con un gran tiempo de anticipación.




**2. El Acoplamiento Océano-Atmósfera y el Transporte de Ekman**

> **El forzante del viento**: El viento que sopla sobre la superficie del mar ejerce una fuerza de fricción conocida como tensión del viento (tau)7. Esta tensión pone en movimiento los primeros metros de la columna de agua.


> **La desviación de Coriolis (Dinámica de Ekman)**: Debido a la rotación de la Tierra, el agua superficial no se mueve en la misma dirección que el viento. En el Hemisferio Sur, el agua se desvía físicamente hacia la izquierda de la dirección del viento (y hacia la derecha en el Hemisferio Norte). La integración vertical de este movimiento genera el Transporte de Ekman, que desplaza inmensas masas de agua superficial a 90° de la dirección del viento






**3. Condiciones Normales vs. El Niño**

> **Normal**: Los vientos alisios soplan con fuerza de este a oeste. El transporte de Ekman acumula agua cálida superficial en el Pacífico Occidental (cerca de Asia), elevando físicamente el nivel del mar allí y hundiendo la termoclina. En nuestras costas sudamericanas (Pacífico Oriental), el agua que se desplaza hacia el oeste genera un "vacío" que fuerza el ascenso de aguas profundas, frías y densas (surgencia o upwelling).

> **El Niño**: Los vientos alisios se debilitan. Sin esa fuerza de empuje, la inmensa "piscina cálida" acumulada en el oeste fluye de regreso hacia Sudamérica en forma de ondas de gravedad profundas (Ondas Kelvin). Al llegar a nuestras costas, el agua cálida superficial suprime la surgencia fría, la termoclina se profundiza y se altera por completo la transferencia de calor y humedad hacia la atmósfera continental
